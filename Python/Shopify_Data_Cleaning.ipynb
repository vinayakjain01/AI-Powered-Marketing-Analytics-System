```python
2. # identified Duplicates
duplicates = df[df.duplicated()]
print("Duplicate rows:", len(duplicates))
```

```python
# Remove Duplicates
df = df.drop_duplicates()
```

```python
3. # Standardise Date Formats
date_cols = ['Date', 'Transaction Timestamp', 'Order Created At', 'Order Updated At']
for col in date_cols:
    # Replace string 'NAN' with actual Nulls and convert to datetime
    df[col] = pd.to_datetime(df[col].replace('NAN', np.nan), errors='coerce', utc=True)
```

```python
# idenifying missing values
df.isnull().sum()
```

```python
# Handling missing value
# - Drop rows missing 'Order ID' (Primary Key) as they cannot be tracked.
df = df.dropna(subset=['Order ID'])
```

```python
df.dtypes
```

```python
# - Fill numeric metrics with 0.0 to prevent calculation breaks.
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(0)
```

```python
# - Fill categorical info with 'unknown' to maintain record count.
string_cols = df.select_dtypes(include=['object']).columns
df[string_cols] = df[string_cols].fillna('unknown')
```

```python
# Normalize String Column 
for col in string_cols:
    df[col] = df[col].astype(str).str.lower().str.strip()
```
